```python
def crea_sala():
    sala = [["L" for posto in range(10)] for fila in range(5)]
    return sala


def mostra_sala(sala):
    print("\n----- SALA CINEMA -----")

    for i, fila in enumerate(sala):
        print(f"Fila {i + 1}: ", end="")

        for posto in fila:
            print(posto, end=" ")

        print()

    print("-----------------------\n")


def prenota(sala, fila, posto):
    # Controllo validità delle coordinate
    if fila < 1 or fila > 5 or posto < 1 or posto > 10:
        return False

    # Conversione delle coordinate negli indici della lista
    fila = fila - 1
    posto = posto - 1

    # Controllo disponibilità del posto
    if sala[fila][posto] == "L":
        sala[fila][posto] = "X"
        return True
    else:
        return False


# Programma principale
sala = crea_sala()

while True:
    mostra_sala(sala)

    fila_input = input("Inserisci la fila (1-5) oppure 'q' per uscire: ")

    if fila_input.lower() == "q":
        break

    posto_input = input("Inserisci il posto (1-10) oppure 'q' per uscire: ")

    if posto_input.lower() == "q":
        break

    # Controllo che gli input siano numeri
    if not fila_input.isdigit() or not posto_input.isdigit():
        print("Coordinate non valide")
        continue

    fila = int(fila_input)
    posto = int(posto_input)

    # Tentativo di prenotazione
    if fila < 1 or fila > 5 or posto < 1 or posto > 10:
        print("Coordinate non valide")
    elif prenota(sala, fila, posto):
        print("Prenotazione effettuata")
    else:
        print("Posto già occupato")

print("Arrivederci!")
```
